```python
# Compare AlphaFold files in a source folder with a list of UniProt accessions
# and copy matching structures to a new folder

import os
import shutil
import csv

# Set source and destination folders
# Update these paths before running the script
source_folder = r'/path/to/source/alphafold/files'
destination_folder = r'/path/to/destination/folder'

# Path to CSV file containing UniProt accessions
csv_file_path = r'/path/to/uniprot_ids.csv'

# Open the CSV file
with open(csv_file_path) as csv_file:
    reader = csv.reader(csv_file)

    # Skip header
    next(reader)

    # Iterate through accession list
    for row in reader:
        accession = f'AF-{row[0]}-F1-model_v3.pdb.gz'
        print(f'Processing accession: {accession}')

        file_path = os.path.join(source_folder, accession)

        if os.path.exists(file_path):
            print(f'Copying file: {accession}')
            shutil.copy(file_path, destination_folder)
        else:
            print(f'File not found: {accession}')
```
